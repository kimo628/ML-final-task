````python
numeric_columns = df.select_dtypes(include=np.number).columns

X = df[numeric_columns]
Mean = df[numeric_columns].mean()
Std = df[numeric_columns].std()
z_score = (X-Mean)/Std

outliers_mask = np.abs(z_score) > 3
outliers_count = outliers_mask.sum()

outliers_count
````

````python
outliers_percentage = (outliers_count / len(df)) * 100

outliers_summary = pd.DataFrame({
    "Outliers Count": outliers_count,
    "Outliers %": outliers_percentage.round(2)
})

outliers_summary
````

````python
# ==============================
# Outlier Treatment
# ==============================

df_clean = df.copy()

outlier_treatment_columns = [
    "CreditScore",
    "Age",
    "Tenure",
    "NumOfProducts"
]

for col in outlier_treatment_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_clean[col] = df_clean[col].clip(
        lower_bound,
        upper_bound
    )

print("Outlier treatment completed.")
````

### Outlier Treatment

Outliers were detected using the Z-score method. To reduce the influence of extreme values, selected numerical features were treated using IQR-based clipping.

The treated dataset (`df_clean`) is used in the modeling phases to ensure that the outlier treatment is part of the final ML pipeline.

# Phase 2 — Regression: Predicting Customer Balance

````python
reg_drop_columns = [
    "Balance",
    "Exited",
    "id",
    "CustomerId",
    "Surname"
]

reg_drop_columns = [
    col for col in reg_drop_columns
]

x_reg = df_clean.drop(reg_drop_columns, axis=1)
y_reg = df_clean["Balance"]

print(x_reg.columns.tolist())
````

````python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
from sklearn.compose import ColumnTransformer

# Train / Test Split
x_train_reg, x_test_reg, y_train_reg, y_test_reg = train_test_split(
    x_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

# Columns
categorical_columns = [
    "Geography",
    "Gender"
]

numeric_columns = [
    col for col in x_reg.columns
    if col not in categorical_columns
]

# Preprocessing
preprocessor_reg = ColumnTransformer(
    transformers=[
        (
            "num",
            MinMaxScaler(),
            numeric_columns
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_columns
        )
    ]
)

# Fit only on training data
x_train_reg_scaled = preprocessor_reg.fit_transform(x_train_reg)

x_test_reg_scaled = preprocessor_reg.transform(x_test_reg)

print("Training shape:", x_train_reg_scaled.shape)
print("Testing shape:", x_test_reg_scaled.shape)
````

````python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

results_reg = []

def evaluate_regression(name, model, X_tr, X_te, y_tr, y_te):
    model.fit(X_tr, y_tr)
    y_pred = model.predict(X_te)

    rmse = np.sqrt(mean_squared_error(y_te, y_pred))
    mae = mean_absolute_error(y_te, y_pred)
    r2 = r2_score(y_te, y_pred)

    results_reg.append({
        "Model": name,
        "RMSE": round(rmse, 2),
        "MAE": round(mae, 2),
        "R2": round(r2, 4)
    })

    print(
        f"{name}: "
        f"RMSE={rmse:.2f} | "
        f"MAE={mae:.2f} | "
        f"R²={r2:.4f}"
    )

    return model


# Linear Regression
linear_model = evaluate_regression(
    "Linear Regression",
    LinearRegression(),
    x_train_reg_scaled,
    x_test_reg_scaled,
    y_train_reg,
    y_test_reg
)

# Ridge Regression
ridge_model = evaluate_regression(
    "Ridge",
    Ridge(alpha=1.0),
    x_train_reg_scaled,
    x_test_reg_scaled,
    y_train_reg,
    y_test_reg
)

# Lasso Regression
lasso_model = evaluate_regression(
    "Lasso",
    Lasso(alpha=1.0,max_iter=10000),
    x_train_reg_scaled,
    x_test_reg_scaled,
    y_train_reg,
    y_test_reg
)

# Polynomial Regression
poly_model = make_pipeline(
    PolynomialFeatures(degree=2, include_bias=False),
    LinearRegression()
)

poly_model = evaluate_regression(
    "Polynomial Regression (deg=2)",
    poly_model,
    x_train_reg_scaled,
    x_test_reg_scaled,
    y_train_reg,
    y_test_reg
)


# Model Comparison
results_reg_df = (
    pd.DataFrame(results_reg)
    .sort_values(by="R2", ascending=False)
    .reset_index(drop=True)
)

print("\n" + "=" * 60)
print("Model Comparison Table - Regression (Target: Balance)")
print("=" * 60)

display(results_reg_df)
````

````python
# ==============================
# Actual vs Predicted Plot
# ==============================

y_pred_poly = poly_model.predict(x_test_reg_scaled)

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test_reg,
    y_pred_poly,
    alpha=0.4
)

plt.plot(
    [y_test_reg.min(), y_test_reg.max()],
    [y_test_reg.min(), y_test_reg.max()],
    linestyle="--"
)

plt.xlabel("Actual Balance")
plt.ylabel("Predicted Balance")
plt.title("Actual vs Predicted Balance — Polynomial Regression")

plt.tight_layout()
````

### Regression Business Interpretation

Polynomial Regression achieved the highest R² among the tested regression models, indicating that it captured some nonlinear relationships in customer balance.

However, the moderate R² shows that customer balance is only partially explained by the available features.

Ridge and Lasso produced performance very close to Linear Regression, suggesting that regularization did not substantially improve predictive performance for this dataset.

# Phase 3 — Classification: Predicting Customer Churn

````python
# Classification Features and Target

classification_drop_columns = [
    "Exited",
    "Balance",
    "id",
    "CustomerId",
    "Surname",
    "HasBalance"
]

classification_drop_columns = [
    col for col in classification_drop_columns
    if col in df.columns
]

x = df_clean.drop(classification_drop_columns, axis=1)
y = df_clean["Exited"]

print("Classification Features:")
print(x.columns.tolist())

print("\nTarget:")
print(y.name)
````

````python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from imblearn.over_sampling import SMOTE

df_model_clean = df_clean.drop(columns=['id', 'CustomerId', 'Surname'])

df_model_clean = pd.get_dummies(df_model_clean, columns=['Geography', 'Gender'], drop_first=True)

x = df_model_clean.drop(columns=['Exited'])
y = df_model_clean['Exited']

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

scaler = MinMaxScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

smote = SMOTE(random_state=42)

x_train_resampled, y_train_resampled = smote.fit_resample(
    x_train_scaled,
    y_train
)
````

## Baseline Classification Models

````python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

from sklearn.model_selection import GridSearchCV
````

````python
log_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

log_model.fit(
    x_train_resampled,
    y_train_resampled
)

y_pred_log = log_model.predict(x_test_scaled)

print("Logistic Regression")
print(classification_report(y_test, y_pred_log))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_log))
````

````python
knn_model = KNeighborsClassifier(
    n_neighbors=5
)

knn_model.fit(
    x_train_resampled,
    y_train_resampled
)

y_pred_knn = knn_model.predict(x_test_scaled)

print("KNN")
print(classification_report(y_test, y_pred_knn))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_knn))
````

````python
tree_model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

tree_model.fit(
    x_train_resampled,
    y_train_resampled
)

y_pred_tree = tree_model.predict(x_test_scaled)

print("Decision Tree")
print(classification_report(y_test, y_pred_tree))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_tree))
````

````python
nb_model = GaussianNB()

nb_model.fit(
    x_train_resampled,
    y_train_resampled
)

y_pred_nb = nb_model.predict(x_test_scaled)

print("Naive Bayes")
print(classification_report(y_test, y_pred_nb))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_nb))
````

````python
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(
    x_train_resampled,
    y_train_resampled
)

y_pred_rf = rf_model.predict(x_test_scaled)

print("Random Forest")
print(classification_report(y_test, y_pred_rf))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))
````

````python
ada_model = AdaBoostClassifier(
    n_estimators=100,
    random_state=42
)

ada_model.fit(
    x_train_resampled,
    y_train_resampled
)

y_pred_ada = ada_model.predict(x_test_scaled)

print("AdaBoost")
print(classification_report(y_test, y_pred_ada))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_ada))
````

````python
models = {
    "Logistic Regression": y_pred_log,
    "SVM": y_pred_svm,
    "KNN": y_pred_knn,
    "Decision Tree": y_pred_tree,
    "Naive Bayes": y_pred_nb,
    "Random Forest": y_pred_rf,
    "AdaBoost": y_pred_ada
}

classification_results = []

for name, y_pred in models.items():

    classification_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1-Score": f1_score(y_test, y_pred)
    })

classification_results_df = pd.DataFrame(
    classification_results
)

classification_results_df.sort_values(
    by="F1-Score",
    ascending=False
).reset_index(drop=True)
````

````python
import time

# Evaluate tuned models on the test set, measuring training time for each.
# This also lets the GUI show the real effect of data volume on training
# time (SVM vs the rest), as called out in the project brief.

tuned_models = {
    "Logistic Regression": log_grid.best_estimator_,
    "SVM": svm_grid.best_estimator_,
    "KNN": knn_grid.best_estimator_,
    "Decision Tree": tree_grid.best_estimator_,
    "Naive Bayes": nb_grid.best_estimator_,
    "Random Forest": rf_grid.best_estimator_,
    "AdaBoost": ada_model
}

tuned_results = []
classification_reports = {}

for name, model in tuned_models.items():

    start_time = time.perf_counter()

    # SVM was trained on the 90,000-sample subset
    if name == "SVM":
        model.fit(x_svm, y_svm)
    else:
        model.fit(x_train_resampled, y_train_resampled)

    training_time = time.perf_counter() - start_time

    y_pred = model.predict(x_test_scaled)

    tuned_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1-Score": f1_score(y_test, y_pred),
        "Training_Time_Sec": round(training_time, 2)
    })

    classification_reports[name] = classification_report(y_test, y_pred)

    print(f"{name} trained in {training_time:.2f} seconds")
````

## Final Tuned Model Comparison

# Phase 4 — Unsupervised Learning: Customer Segmentation

````python
# ==============================
# Unsupervised Learning
# Customer Segmentation
# ==============================

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Select relevant features for customer segmentation
segmentation_features = [
    "Tenure",
    "Balance",
    "NumOfProducts",
    "IsActiveMember",
    "EstimatedSalary",
    "CreditScore",
    "Age"
]

X_seg = df_clean[segmentation_features].copy()

print("Segmentation Features:")
print(segmentation_features)

print("\nShape:")
print(X_seg.shape)

print("\nMissing Values:")
print(X_seg.isnull().sum())
````

````python
# Scale the features
scaler_seg = StandardScaler()

X_seg_scaled = scaler_seg.fit_transform(X_seg)

print("Scaled data shape:", X_seg_scaled.shape)
````

## Silhouette Score — Selecting Number of Clusters

````python
# ==============================
# Silhouette Score
# ==============================

silhouette_scores = []
k_values = range(2, 11)

sample_size = 80000

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_seg_scaled)
    
    score = silhouette_score(
        X_seg_scaled,
        labels,
        sample_size=sample_size,
        random_state=42
    )
    
    silhouette_scores.append(score)
    print(f"K = {k} -> Silhouette = {score:.4f}")

silhouette_results = pd.DataFrame({
    "K": list(k_values),
    "Silhouette Score": silhouette_scores
})

print("\nFinal Results:")
print(silhouette_results.round(4))
````

````python
plt.figure(figsize=(8, 5))
plt.plot(k_values, silhouette_scores, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for KMeans")
plt.xticks(k_values)
````

## Elbow Method

````python
# ==============================
# Elbow Method
# ==============================

inertia = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(X_seg_scaled)
    inertia.append(kmeans.inertia_)

elbow_results = pd.DataFrame({
    "K": list(k_values),
    "Inertia": inertia
})

print(elbow_results.round(2))
````

````python
plt.figure(figsize=(8, 5))
plt.plot(k_values, inertia, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.title("Elbow Method for KMeans")

plt.xticks(k_values)
````

## Final KMeans Segmentation

````python
# ==============================
# Final KMeans Model
# ==============================

best_k = 4

kmeans_final = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

df_clean["KMeans_Cluster"] = kmeans_final.fit_predict(X_seg_scaled)

print("KMeans clustering completed.")
print("\nCluster Counts:")
print(df_clean["KMeans_Cluster"].value_counts().sort_index())
````

## Cluster Profiling

````python
# ==============================
# KMeans Cluster Profiling
# ==============================

cluster_profile = df_clean.groupby("KMeans_Cluster")[segmentation_features].mean().round(2)

cluster_sizes = (
    df_clean["KMeans_Cluster"]
    .value_counts()
    .sort_index()
)

cluster_profile["Customer_Count"] = cluster_sizes.values

cluster_profile["Percentage"] = (
    cluster_sizes.values / len(df_clean) * 100
).round(2)

print(cluster_profile)
````

### Business Interpretation of Customer Segments

Four broad customer profiles were identified using KMeans.

The segments differ mainly in customer activity, balance, age, number of products, and observed churn rate.

The relatively low Silhouette Score indicates that the clusters should be interpreted as broad behavioral profiles rather than perfectly separated customer groups.

These profiles can still support differentiated retention and customer-management strategies.
